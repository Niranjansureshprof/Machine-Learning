## Step 1: Import libraries

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
```

## Step 2: Import Dataset

```python
claimants_data=pd.read_csv("Claimants.csv")
claimants_data.head()
```

## Step 3: Data Understanding

```python
claimants_data.shape
```

```python
claimants_data.dtypes
```

```python
claimants_data.describe()
```

```python
claimants_data.isna().sum()
```

```python
claimants_data.dropna(inplace=True)
```

```python
claimants_data.shape
```

## Step 4: Data Preparation

```python
X=claimants_data[["CLMSEX","CLMINSUR","SEATBELT","CLMAGE","LOSS"]]
y=claimants_data["ATTORNEY"]
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,shuffle=True,random_state=37)
```

## Step 5: Model Building

```python
rf_classifier=RandomForestClassifier(criterion= 'gini', max_depth= 5, n_estimators= 125)
```

## Step 6: Model Training

```python
rf_classifier.fit(X_train,y_train)
```

## Step 7: Model Testing

### Testing with training data

```python
y_train_pred=rf_classifier.predict(X_train)
y_train_pred
```

### Testing with testing data

```python
y_test_pred=rf_classifier.predict(X_test)
y_test_pred
```

## Step 8: Model Evaluation

### Training Data Evaluation Metrics

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report,roc_curve,roc_auc_score
```

```python
accuracy_score(y_train,y_train_pred)
```

```python
confusion_matrix(y_train,y_train_pred)
```

```python
classification_report(y_train,y_train_pred)
```

```python
fpr,tpr,thresholds=roc_curve(y_train,rf_classifier.predict_proba(X_train)[:,1])
auc_score=roc_auc_score(y_train,y_train_pred)

plt.figure(figsize=(7, 5))
plt.plot(fpr, tpr, color='blue', lw=2, label=f'ROC curve (AUC = {auc_score:.2f})')
plt.plot([0, 1], [0, 1], color='gray', linestyle='--', lw=2, label='Random Guessing (AUC = 0.50)')

# Formatting
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate (FPR)')
plt.ylabel('True Positive Rate (TPR)')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc="lower right")
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

### Testing Data Evaluation Metrics

```python
accuracy_score(y_test,y_test_pred)
```

```python
confusion_matrix(y_test,y_test_pred)
```

```python
classification_report(y_test,y_test_pred)
```

```python
fpr,tpr,thresholds=roc_curve(y_test,rf_classifier.predict_proba(X_test)[:,1])
auc_score=roc_auc_score(y_test,y_test_pred)

plt.figure(figsize=(7, 5))
plt.plot(fpr, tpr, color='blue', lw=2, label=f'ROC curve (AUC = {auc_score:.2f})')
plt.plot([0, 1], [0, 1], color='gray', linestyle='--', lw=2, label='Random Guessing (AUC = 0.50)')

# Formatting
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate (FPR)')
plt.ylabel('True Positive Rate (TPR)')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc="lower right")
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```
