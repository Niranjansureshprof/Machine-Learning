##  Step 1: Import Libraries

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
```

## Step 2: Import Dataset

```python
claimants_data=pd.read_csv("Claimants.csv")
claimants_data
```

## Step 4: Data Preparation

#### 4.1 Data Cleaning

```python
claimants_data=claimants_data.dropna()
```

```python
claimants_data.isna().sum()
```

```python
claimants_data.shape
```

#### 4.2 Separate your inputs and output

```python
X=claimants_data.drop(labels=["CASENUM","ATTORNEY"],axis=1)
X
#Dropping CASENUM, as it's not gonna make any sense in prediction and ATTORNEY is output
```

```python
y=claimants_data["ATTORNEY"]
```

## Step 5: Model Building

```python
claimants_model=LogisticRegression()
```

## Step 6: Model Training 

```python
claimants_model.fit(X,y)
```

## Step 7: Model Testing 

```python
y_pred=claimants_model.predict(X)
claimants_data["Predicted Output"]=y_pred
claimants_data["Predicted Output"]
```

## Step 8: Model Evaluation 

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_true=y,y_pred=y_pred)
```

## Step 9: Model Deployment

```python
from pickle import dump
dump(claimants_model,open("claimants.pkl","wb"))
```
