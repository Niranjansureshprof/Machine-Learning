## Step 1: Import Libraries

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error
```

## Step 2: Import Dataset

```python
cars_data=pd.read_csv("Cars.csv")
cars_data.head()
```

## Step 3: Data Understanding

```python
cars_data.dtypes
```

```python
cars_data.describe()
```

```python
cars_data.isna().sum()
```

```python
cars_data.shape
```

## Step 4: Data Preparation

```python
X=cars_data[["HP","SP","WT","VOL"]]
X
```

```python
y=cars_data["MPG"]
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,shuffle=True,random_state=41)
```

```python
X_train
```

```python
X_test
```

```python
y_train
```

```python
y_test
```

## Step 5: Model Building

```python
cars_model=LinearRegression()
```

## Step 6: Model Training

```python
cars_model.fit(X_train,y_train)
```

## Step 7: Model Testing

Testing Training data

```python
y_pred_train=cars_model.predict(X_train)
y_pred_train
```

Testing test data

```python
y_pred_test=cars_model.predict(X_test)
y_pred_test
```

## Step 8: Model Evaluation

Training data error

```python
training_error=y_train-y_pred_train
training_error
```

```python
mean_absolute_error(y_train,y_pred_train)
```

Teesting data error

```python
testing_error=y_test-y_pred_test
testing_error
```

```python
mean_absolute_error(y_test,y_pred_test)
```
