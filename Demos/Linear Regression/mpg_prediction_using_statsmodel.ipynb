```python
import pandas as pd
import statsmodels.formula.api as smf
from sklearn.model_selection import train_test_split
```

```python
cars_data=pd.read_csv("Cars.csv")
```

## Step 4: Data preparation

```python
train,test=train_test_split(cars_data,test_size=0.2,shuffle=True,random_state=11)
```

## Step 5: Model Building

```python
cars_statsmodel=smf.ols(formula="MPG~HP+SP+WT+VOL",data=cars_data)
```

## Step 6: Model Training

```python
cars_statsmodel=cars_statsmodel.fit()
```

## Step 7: Model Testing

### Testing on training data

```python
tain_pred=cars_statsmodel.predict(train)
```

### Testing on test data

```python
test_pred=cars_statsmodel.predict(test)
```
